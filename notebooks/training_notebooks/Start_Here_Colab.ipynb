```python
from pathlib import Path
import zipfile
archive_path = Path("/content/AI_for_Everyone_Ready_Datasets.zip")
with zipfile.ZipFile(archive_path) as archive:
    archive.extractall("/content")
```

```python
import json
import numpy as np
import pandas as pd
PROJECT_DIR = Path("/content/AI_for_Everyone_Ready_Datasets/Project_03")
DATA = PROJECT_DIR / "data"
(PROJECT_DIR / "models").mkdir(exist_ok=True)
(PROJECT_DIR / "reports").mkdir(exist_ok=True)
meta = json.loads((DATA / "meta.json").read_text())
main = pd.read_csv(DATA / "main.csv")
feedback = pd.read_csv(DATA / "feedback.csv")
images = np.load(DATA / "images.npz", allow_pickle=False)
print(main.groupby("split").size())
print({name: images[name].shape for name in images.files})
main.head()
```

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, min(6, len(images["x_train"])), figsize=(10, 2))
for i, ax in enumerate(axes):
    image = images["x_train"][i]
    ax.imshow(image.squeeze(), cmap="gray" if image.shape[-1] == 1 else None)
    ax.set_title(meta["image_classes"][int(images["y_train"][i])])
    ax.axis("off")
plt.tight_layout()
```
